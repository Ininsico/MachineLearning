# Q6 - Prime number check

Tests n for primality by trial division.

Run the cell, then answer each prompt - press Enter to accept the default.

<sub>Colab notebook generated from `homework/q6.py`</sub>

```python
n = int(input("Enter a number (default 29): ").strip() or 29)
is_prime = True
```

```python
if n < 2:
    is_prime = False
else:
    i = 2
    while i * i <= n:
        if n % i == 0:
            is_prime = False
            break
        i += 1

if is_prime:
    print(f"{n} is a prime number.")
else:
    print(f"{n} is not a prime number.")
```
