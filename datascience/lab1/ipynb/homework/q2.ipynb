# Q2 - Largest of three numbers

Reads three numbers and prints the largest.

Run the cell, then answer each prompt - press Enter to accept the default.

<sub>Colab notebook generated from `homework/q2.py`</sub>

```python
a = float(input("Enter the first number (default 12): ").strip() or 12)
b = float(input("Enter the second number (default 45): ").strip() or 45)
c = float(input("Enter the third number (default 33): ").strip() or 33)
```

```python
if a >= b and a >= c:
    largest = a
elif b >= a and b >= c:
    largest = b
else:
    largest = c

print(f"Numbers entered: {a:g}, {b:g}, {c:g}")
print(f"The largest number is: {largest:g}")
```
