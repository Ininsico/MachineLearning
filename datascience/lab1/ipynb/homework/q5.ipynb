# Q5 - Factorial

Computes n! with a loop; negative input is rejected.

Run the cell, then answer each prompt - press Enter to accept the default.

<sub>Colab notebook generated from `homework/q5.py`</sub>

```python
n = int(input("Enter a number (default 5): ").strip() or 5)
```

```python
if n < 0:
    print("Factorial is not defined for negative numbers.")
else:
    factorial = 1
    for i in range(1, n + 1):
        factorial *= i

    print(f"{n}! = {factorial}")
```
