Prints the first n Fibonacci numbers.

Run the cell, then answer each prompt - press Enter to accept the default.

```python
n = int(input("How many terms? (default 10): ").strip() or 10)
a = 0
b = 1
```

```python
print("Fibonacci Series:", end=" ")
for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
print()
```
