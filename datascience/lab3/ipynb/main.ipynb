```python
from pathlib import Path
import pandas as pd
file_path = Path("main.xlsx")
students = pd.read_excel(file_path, sheet_name="Sheet1")
```

```python
print(students.head())
print(students.info())
print(students.shape)
```

```python
marks = students.drop(columns=["S#", "Reg. #", "Name"]).apply(pd.to_numeric, errors="coerce")
print(marks.mean())
```
