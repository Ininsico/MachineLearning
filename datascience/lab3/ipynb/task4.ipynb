```python
from pathlib import Path

files_needed = ["neo.csv"]
missing = [f for f in files_needed if not Path(f).exists()]
if missing:
    try:
        from google.colab import files
        print("Upload:", ", ".join(missing))
        files.upload()   # select the file(s) in the picker
    except ImportError:
        print("Not running in Colab; place the dataset next to this notebook.")
```

```python
import threading
from http.server import SimpleHTTPRequestHandler, ThreadingHTTPServer
from pathlib import Path

try:
    server = ThreadingHTTPServer(("localhost", 8000), SimpleHTTPRequestHandler)
    threading.Thread(target=server.serve_forever, daemon=True).start()
    print("serving http://localhost:8000 from", Path.cwd())
except OSError:
    print("port 8000 already in use - assuming a server is already running")
```

```python
import pandas as pd
data_py = pd.read_csv("http://localhost:8000/neo.csv")
```
